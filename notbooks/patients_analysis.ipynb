```python
df = pd.read_csv("../data/patients.csv", sep=",", header=0, engine="python")
df = df.iloc[:, 0].str.split(",", expand=True)
df.columns = ["patientID", "Age", "Sex", "Bloodpressure", "Cholesterol", "Daignosis"]
print(df.head())
```

```python
# Run basic descriptive statistics using:
print(df.describe())
print(df["Sex"].value_counts())
print(df["Daignosis"].value_counts())
```

```python
import matplotlib.pyplot as plt

plt.plot(df["Age"], df["Cholesterol"], marker="o")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.title("Cholesterol vs Age")
plt.show()
```
